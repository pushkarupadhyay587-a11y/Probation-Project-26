```python

import numpy as np
```

```python
arr = np.random.randint(1,101,size=10)
print(arr)
```

```python
# 1.a:
max_element = np.max(arr)
print("max element:",max_element)
```

```python
# 1.b :
min_element = np.min(arr)
print("minimum element :",min_element)
```

```python
#  1.c :
mean = np.mean(arr)
print("mean : " ,mean)
```

```python
# 1.d :
s = np.sum(arr)
print("sum of the elemnents :", s)
```

```python
# 1.e :
max_index = np.argmax(arr)
print("index of the max element : ",max_index)
```

```python
# 1.f :
min_index  = np.argmin(arr)
print("index of minimum element: ",min_index)
```

```python
arr=np.sort(arr)
print(arr)
```
