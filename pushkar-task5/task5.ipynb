# checking outliers

```python
q1 = df["avg cost (two people)"].quantile(0.25)
q3 = df["avg cost (two people)"].quantile(0.75)

iqr = q3-q1 

lower_bound = q1-1.5*iqr
upper_bound = q3 + 1.5*iqr 

print(lower_bound," - ",upper_bound)
```

```python
num_outliers = ((df["avg cost (two people)"] <lower_bound )| (df["avg cost (two people)"]>upper_bound)).sum()
```

```python
num_outliers
```

```python
df["avg cost (two people)"] = df["avg cost (two people)"].clip(0,upper_bound)
df["avg cost (two people)"]
```

```python
q1 = df["num of ratings"].quantile(0.25)
q3 = df["num of ratings"].quantile(0.75)

iqr = q3-q1 

lower_bound_rating = q1-1.5*iqr
upper_bound_rating = q3 + 1.5*iqr 

print(lower_bound_rating," - ",upper_bound_rating)
```

```python
df["num of ratings"] = df["num of ratings"].clip(0,upper_bound_rating)
df["num of ratings"]
```

```python
import codecs

def fix_text(name):
    
    fixed = name
    for _ in range(10):
        new_fixed = codecs.decode(fixed.encode("latin1"), "utf-8", "ignore")
        if new_fixed == fixed:
            break
        fixed = new_fixed
    return fixed
```

```python
df["restaurant name"] = df["restaurant name"].apply(fix_text)
df["restaurant name"].head(10)
```
